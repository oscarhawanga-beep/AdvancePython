```python
# Create a pre poaching data frame
# Display the first three rows

post_poaching = elephant.query('period =="2005-13"')
post_poaching.head(3)
```

```python
# Calculate the average tusk length for the pre-poaching period.
# Repeat for the post-recovery period
    
print("Pre :",pre_poaching['tusk_length'].mean())
print("Post :",post_poaching['tusk_length'].mean())
```
