```python
# we import two additional module from python import seaborn as sns
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.relplot("gender","itemprice",data=data)
```

```python
sns.countplot(x="gender", data=data)
```

```python
ax = sns.countplot(x="city", hue="payment", data=data, color="green")

plt.title("Products pers different cities")
plt.ylabel("Number of products")
plt.xlabel("Name of city")
plt.xticks(rotation=45)
for container in ax.containers:
    ax.bar_label(container)
plt.show()
```

```python
data["city"].value_counts().plot.pie(autopct="1%.0f%%")
```

```python
sns.set(style="whitegrid")
```

```python
sns.swarmplot(x="gender", y="customersalary", data=data,palette="Set2",hue="payment")
```
