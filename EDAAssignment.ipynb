```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv("datasets/bank.csv")
data
```

```python
data.describe()
```

```python
data.groupby(
    ["Gender", 
     "Education",
     "Married"]
     )
.size()
```

```python
data.isnull().sum()
```

```python
data["Self_Employed"] = data["Self_Employed"].fillna("Unknown")
```

```python
sns.countplot(
    "Gender",
).show()
plt.text()
```

```python
data.isnull().sum()
```

```python
sns.scatterplot(
    data=data,
    x="ApplicantIncome",
    y="LoanAmount",
    hue="Gender",
    palette="Set2",
)

plt.title("Applicant Income vs Loan Amount by Gender")
plt.xlabel("Applicant Income")
plt.ylabel("Loan Amount")
plt.tight_layout()
plt.show()
```

```python
sns.swarmplot(
    data=data,
    x="Gender",
    y="LoanAmount",
    hue="Gender",
    palette="Set2",
    size=5,
)

plt.title("Loan Amount Distribution by Gender")
plt.xlabel("Gender")
plt.ylabel("Loan Amount")
plt.tight_layout()
plt.show()
```
