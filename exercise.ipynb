```python
#Assign a color column.
#Assign a radius column.
#Show the result.
import pandas as pd
df=pd.DataFrame()
df['colour']=['red','blue','orange','green','yellow']
df['radius']=[2,4,6,8,10]
df





```

```python
#Calculate a diameter column.
#Output the result.
df['diameter']=df['radius']* 2
df['progress']=df['diameter']+df['radius']*3.142
df
```

```python
#smallest diameter
df['radius'].sum()
```

```python
#-1 for last row, 0 for first
df.iloc[1]
```

```python
# Matplotlib exercise
import matplotlib.pyplot as plt
x=df['colour']
y=df['radius']
plt.bar(x,y, color = 'green')
plt.plot(x,y,color='black')
plt.ylabel("Radius")
plt.xlabel('Colours')
plt.title("New Data Scientist in Town")
plt.show()
```
